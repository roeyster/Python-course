```python
salary_year_median = df['salary_year_avg'].median()
salary_hour_median = df['salary_hour_avg'].median()

print(salary_year_median)
print(salary_hour_median)
```

```python
df_fill = df

df_fill['salary_year_avg'] = df_fill['salary_year_avg'].fillna(salary_year_median)
df_fill['salary_hour_avg'] = df_fill['salary_hour_avg'].fillna(salary_hour_median)
```

```python
df_uniqe = df_fill

df_uniqe = df_uniqe.drop_duplicates()

df_uniqe = df_uniqe.drop_duplicates(subset=['job_title', 'company_name', 'job_work_from_home', 'job_location'])

print('length before: ', len(df_fill))
print('length after: ', len(df_uniqe))
print('coulmns removed: ', len(df_fill)-len(df_uniqe))
```
