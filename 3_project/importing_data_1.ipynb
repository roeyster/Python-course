```python
import pandas as pd
import requests
from lxml import html

# Step 1: Fetch the webpage
url = "https://he.wikipedia.org/wiki/%D7%94%D7%9B%D7%95%D7%9B%D7%91_%D7%94%D7%91%D7%90_%D7%9C%D7%90%D7%99%D7%A8%D7%95%D7%95%D7%99%D7%96%D7%99%D7%95%D7%9F_2025"  # Replace with actual URL
response = requests.get(url)
tree = html.fromstring(response.content)  # Parse HTML

# Step 2: Find all tables manually using lxml
tables = tree.xpath("//table")  # List of all tables
filtered_tables = []  # List to store valid tables

```

```python
# Step 3: Process each table
for table in tables:
    data = []
    headers = [th.text_content().strip() for th in table.xpath(".//th")]  # Extract headers
    
    first_img_src = None  # Store the first image found

    # Step 4: Extract table rows
    for row in table.xpath(".//tr"):
        row_data = []
        for cell in row.xpath(".//td"):
            img = cell.xpath(".//img/@src")  # Get image src if exists
            if img:
                img_src = img[0]  # Extract first image source
                if first_img_src is None:
                    first_img_src = img_src  # Save first image as "Yes"
                
                # Compare to the first image
                row_data.append("Yes" if img_src == first_img_src else "No")
            else:
                row_data.append(cell.text_content().strip())  # Extract normal text

        if row_data:
            data.append(row_data)

   

```

```python
 # Step 5: Convert to DataFrame & Filter by "שם השיר"
    if data:
        df = pd.DataFrame(data, columns=headers)  # Create DataFrame
        if "שם השיר" in df.columns:  # Check if column exists
            filtered_tables.append(df)
```

```python
# Step 6: Print results
for i, table in enumerate(filtered_tables):
    print(f"\nTable {i}:\n", table)
```
